```python
rootfile_name = 'ps_output.root'

rootFile = uproot.open(rootfile_name)
# Printing keys with associated objects
print(rootFile.classnames())
```

```python
psData = rootFile['histPS_0_0;2'].to_numpy()
psVals = psData[0]
psEdges = psData[1]

psCenters = (psEdges[:-1] + psEdges[1:]) / 2

fig, ax = plt.subplots(figsize=(6,5))
ax.plot(psCenters, psVals)
ax.set_ylabel('Power [W]')
ax.set_xlabel('Frequency [Hz]')
```
